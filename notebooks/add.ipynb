```python
a = 3  # Amount of numbers to add
numbers = [10, 20, 30]

assert len(numbers) == a, "The amount of numbers must be equal to a"
```

```python
som = sum(numbers)
print(f"De som van {numbers} = {som}")  
```

```python
def add_numbers(num_list):
    """Function to add numbers in a list."""
    return sum(num_list)
```

```python
list_of_numbers = [1, 2, 3, 4, 5]
result = add_numbers(list_of_numbers)
print(f"The sum of {list_of_numbers} is {result}")
```

```python
list1 = [1, 2, 3]
list2 = [4, 5, 6]
```

```python
print(add_numbers(list1))  # Output: 6
print(add_numbers(list2))  # Output: 15
```
